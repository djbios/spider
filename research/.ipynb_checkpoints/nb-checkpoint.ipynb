```python
from ikpy.chain import Chain
from ikpy.link import OriginLink, URDFLink
import math

leg1 = Chain(name='leg1', links=[
    OriginLink(),
    URDFLink(
      name="hip",
      origin_translation=[10, -10, 0],
      origin_orientation=[0, 0, math.radians(-90)],
      rotation=[0, 0, 1],
    ),
    URDFLink(
      name="knee",
      origin_translation=[10, 0, 0],
      origin_orientation=[0, math.radians(180), 0],
      rotation=[0, -1, 0],
    ),
    URDFLink(
      name="ankle",
      origin_translation=[0, 0, 10],
      origin_orientation=[0, math.radians(-90), 0],
      rotation=[0, 1, 0],
    ),
    URDFLink(
      name="pyatka",
      origin_translation=[10, 0, 0],
      origin_orientation=[0, 0, 0],
      rotation=[1, 0, 0],
    )
])
leg1.inverse_kinematics([100,-100,100])
```

```python
import matplotlib.pyplot
from mpl_toolkits.mplot3d import Axes3D
ax = matplotlib.pyplot.figure().add_subplot(111, projection='3d')

#leg1.plot(my_chain.inverse_kinematics([10, 10, 10]), ax)
#leg1.plot([math.radians(0),math.radians(90),math.radians(90), math.radians(90), math.radians(0)], ax)
leg1.plot([math.radians(0),math.radians(23),math.radians(46), math.radians(10), math.radians(0)], ax)
matplotlib.pyplot.show()
```
